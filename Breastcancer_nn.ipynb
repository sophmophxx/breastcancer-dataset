# 1. Setup and Dependencies

```python
import numpy as np
import pandas as pd
import sklearn
import torch
import matplotlib

import matplotlib.pyplot as plt


from sklearn.datasets import load_breast_cancer
```

```python
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Scikit-learn:", sklearn.__version__)
print("PyTorch:", torch.__version__)
print("Matplotlib:", matplotlib.__version__)
```

## 2. Data Loading and Exploration

```python
bc = load_breast_cancer(as_frame=True)
X = bc.data
y = pd.Series(bc.target).map({0: "malignant", 1: "benign"}).rename("status")
df = pd.concat([X, y], axis=1)
```

### 2.1 Dataset Overview

```python
# Dataset dimensions
print("Dataset shape:",df.shape)

# First five observations
display(df.head())

# Dataset information
df.info()
```

### 2.3 Target Distribution

```python
df["status"].value_counts()
```

```python
df["status"].value_counts(normalize=True)
```

```python
class_counts = df["status"].value_counts()

plt.bar(class_counts.index, class_counts.values)

plt.xlabel("Diagnosis")
plt.ylabel("Number of Samples")
plt.title("Distribution of Target Class")

plt.show()
```

```python
class_counts = df["status"].value_counts(normalize=True) * 100

plt.bar(class_counts.index, class_counts.values)

plt.xlabel("Diagnosis")
plt.ylabel("Percentage (%)")
plt.title("Distribution of Target Classes")

plt.show()
```

### 2.4 Feature Analysis

```python
df.describe()
```

```python
plt.hist(df["mean radius"], bins=20)

plt.xlabel("Mean radius")
plt.ylabel("Frequency")
plt.title("Distribution of Mean Radius")

plt.show()
```

```python
plt.hist(
    df[df["status"] == "benign"]["mean radius"],
    bins=20,
    alpha=0.5,
    label="Benign",
)

plt.hist(
    df[df["status"] == "malignant"]["mean radius"],
    bins=20,
    alpha=0.5,
    label="Malignant",
)

plt.xlabel("Mean Radius")
plt.ylabel("Frequency")
plt.title("Mean Radius Distribution by Diagnosis")
plt.legend()

plt.show()
```

### 2.5 Feature Correlation Analysis

```python
features = df.drop(columns=["status"])

correlation_matrix = features.corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 10))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Pearson Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```
